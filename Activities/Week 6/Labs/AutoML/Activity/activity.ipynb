```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(df, test_size=1000, random_state=42, stratify=df['Churn'])

# Training set for the model building and test set used once at the end to check the performance
```

Question: What type of ML problem is this?

This is a classification problem

```python
from pycaret.classification import *

# Initialize setup
clf_setup = setup(
    data=train_df, 
    target='Churn', 
    session_id=42,
    train_size=0.8
)

# Compare models
best_models = compare_models(n_select=3)
```

```python
# We will analyze teh results 
# Look at metrics like Accuracy, Precision, Recall, F1-score, AUC
plot_model(best_models[0], plot='confusion_matrix')
```
