```python
# Dependencies
import requests
import json
```

```python
# Base URL for GET requests to retrieve number/date facts
base_url = 'http://numbersapi.com/'
```

```python
# Ask the user what kind of data they would like to search for
number = input('Enter a number you would like facts on! ')
fact = input('Which fact type would you like? Date, Math, Year, or trivia ')
```

```python
# Create code to return a number fact
url = f'{base_url}{number}/{fact}?json'
response = requests.get(url)
data= response.json()
```

```python
print(f'Fun Fact on The Number {data['number']}: {data['text']}')
```
