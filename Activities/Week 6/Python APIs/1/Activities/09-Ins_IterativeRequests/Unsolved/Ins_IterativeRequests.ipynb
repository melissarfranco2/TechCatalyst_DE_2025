```python
# Dependencies
import random
import json
import requests
```

```python
# Let's get the JSON for 100 posts sequentially.
url = "http://jsonplaceholder.typicode.com/posts/"
```

```python
# Create an empty list to store the responses
responses = []
```

```python
# Create random indices representing
# a user's choice of posts
random_indices = random.sample(list(range(1,100)),10)
```

```python
# Make a request for each of the indices
for index in random_indices:
    index_request = requests.get(url+str(index))
    responses.append(index_request.json())
```

```python
# Now we have 10 post objects, 
# which we got by making 100 requests to the API.
print(len(responses))
```
