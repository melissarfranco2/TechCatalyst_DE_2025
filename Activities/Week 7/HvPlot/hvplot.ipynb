```python
import pandas as pd
import matplotlib.pyplot as plt
plt.rc("figure", figsize=(16, 5))
plt.style.use('https://github.com/dhaitz/matplotlib-stylesheets/raw/master/pitayasmoothie-light.mplstyle')
```

```python
df = pd.read_csv("./datasets/closing_price.csv",
                 index_col='Date',
                 parse_dates=True)
```

```python
df.info()
```

```python
df.head()
```

```python
df.plot(title='Stock Prices Over Time',
        xlabel='Date',
        ylabel='Price')
plt.show()
```

```python
df.plot(title='Stock Prices Over Time',
        xlabel='Date',
        ylabel='Price',
        subplots=True)
plt.show()
```

```python
df.plot(title='Stock Prices Over Time',
        xlabel='Date',
        ylabel='Price',
        subplots=True,
        sharex=False)
plt.show()
```

```python
df.div(df.iloc[0]).plot()
```

```python
df_normalized = df.div(df.iloc[0])
df_normalized.plot(
    title='Normalized Stock Data',
    ylabel='Normalized Price',
)
```

```python
df.pct_change().plot()
```

```python
import hvplot.pandas
```

```python
df.hvplot(title='Stock Prices Over Time',
        xlabel='Date',
        ylabel='Price')
```

```python
df.hvplot(title='Stock Prices Over Time',
        ylabel='Price',
        xlabel='Dates',
        subplots=True,
        sharex=False,
        sharey=True)
```
